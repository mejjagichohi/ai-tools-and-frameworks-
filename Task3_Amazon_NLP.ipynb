```python
# Load the model we just downloaded
nlp = spacy.load("en_core_web_sm")
```

```python
# Sample Data
reviews = [
    "I absolutely love the Sony WH-1000XM5 headphones! The noise cancellation is fantastic.",
    "The shipping for my generic blender was terrible and the box was crushed.",
    "Apple usually makes great products, but this iPhone case feels cheap."]
```

```python
# Rule-Based Sentiment Function
def get_rule_based_sentiment(doc):
    positive_words = {"love", "fantastic", "great", "good", "amazing", "excellent"}
    negative_words = {"terrible", "crushed", "bad", "cheap", "poor", "hate"}

    score = 0
    for token in doc:
        if token.lemma_.lower() in positive_words:
            score += 1
        elif token.lemma_.lower() in negative_words:
            score -= 1

    if score > 0: return "Positive"
    elif score < 0: return "Negative"
    else: return "Neutral"

print("\n--- NLP Analysis Results ---")

for text in reviews:
    doc = nlp(text)
```

```python
# Extract Entities (Brands/Products)
entities = [(ent.text, ent.label_) for ent in doc.ents]

# Analyze Sentiment
sentiment = get_rule_based_sentiment(doc)

print(f"\nReview: {text}")
print(f"Entities Found: {entities}")
print(f"Sentiment: {sentiment}")
```
